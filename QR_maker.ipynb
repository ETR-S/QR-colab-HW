## QR Code Maker

Run the installation cell first, then run the program and enter a link.

- Enter `y` to display the QR code and save it as `qrcode.png` in the Colab runtime.
- Enter `n` to display it without saving.
- Uppercase answers and surrounding spaces are accepted.

```python
%pip install "qrcode[pil]"
```

```python
import qrcode
url = input("enter a link: ").strip()
path = "qrcode.png"
qr = qrcode.QRCode()
qr.add_data(url)
img = qr.make_image()
the_user_input = input("do you want to save the qrcode? y/n").strip().lower()
if the_user_input == "y":
    display(img)
    img.save(path)
    print("done!")
elif the_user_input == "n":
    display(img)
    print("done!")
else:
    print("!!! the only acceptable inputs are y or n !!!")
```
